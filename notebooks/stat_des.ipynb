## PRE PROCESS AI GPR

```python

import pandas as pd
from pathlib import Path



# Supprimer la ligne d'en-tête supplémentaire
ai = ai.drop(index="Date", errors="ignore")

# Conversion des dates
ai.index = pd.to_datetime(ai.index, errors="coerce")
ai = ai.loc[ai.index.notna()].copy()
ai.index.name = "date"

# Conversion des variables en numérique
ai = ai.apply(pd.to_numeric, errors="coerce")

# 2. Pays étudiés
countries = {
    "France": "FRA",
    "Germany": "DEU",
    "Italy": "ITA",
    "Spain": "ESP",
    "USA": "USA"
}

roles = ["all", "initiator", "respondent", "spillover"]

# 3. Sélection et restructuration
frames = []

for country_name, iso3 in countries.items():
    columns = [f"{country_name}_{role}" for role in roles]

    country_ai = ai[columns].copy()

    country_ai.columns = [
        "gpr_country",
        "gpr_initiator",
        "gpr_respondent",
        "gpr_spillover"
    ]

    country_ai["country"] = iso3
    country_ai["gpr_global"] = ai["GPR_AI"]

    frames.append(country_ai.reset_index())

panel = pd.concat(frames, ignore_index=True)

# 4. Période d'étude
panel = panel.loc[
    (panel["date"] >= "2000-01-01")
    & (panel["date"] <= "2025-12-01")
]

panel = panel.sort_values(
    ["country", "date"]
).reset_index(drop=True)

# 5. Contrôles
print("Dimensions :", panel.shape)
print("\nObservations par pays :")
print(panel.groupby("country").size())

print("\nValeurs manquantes :")
print(panel.isna().sum())

print("\nAperçu :")
print(panel.head())

# 6. Export
output = Path("C:\\Users\\werid\\Desktop\\projets_personnels\\GEOPOLITIC-RISK-ANALYSIS\\data\\processed\\ai_gpr_country_monthly.csv")
output.parent.mkdir(parents=True, exist_ok=True)

panel.to_csv(output, index=False)
print("\nFichier exporté :", output)
```

```python

import numpy as np

components_sum = (
    panel["gpr_initiator"]
    + panel["gpr_respondent"]
    + panel["gpr_spillover"]
)

difference = (
    panel["gpr_country"] - components_sum
).abs()

print("Écart maximal :", difference.max())
print("Écart moyen :", difference.mean())

print(
    "Décomposition cohérente :",
    np.allclose(
        panel["gpr_country"],
        components_sum,
        atol=1e-3,
        rtol=0
    )
)
```
